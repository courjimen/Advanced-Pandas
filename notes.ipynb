### Handling Multiple DataFrames
- `df.merge()`: similar to JOIN in SQL. Looks for common columns or keys. Makes data set wider/add columns
- `df.join()`: similar to JOIN in SQL. Combines DF on index. Makes data set wider/add columns
- `df.concat()`: similar to a UNION in SQL. Add rows. Stacking DF on top of ea other. Makes dataset longer. 
- `df.append()`: deprecated, don't use. USE CONCAT instead

# .merge()

```python
left_df = pd.DataFrame(
    {
        'Courses': ["Spark","PySpark","Python","pandas","Java"],
        'Fee' : [20000,25000,30000,24000,40000],
        'Duration':['30day','40days','60days','55days','50days']
    })

right_df = pd.DataFrame(
    {
            'Courses': ["Java","PySpark","Python","pandas","Hyperion","html"],
            'Fee': [20000,25000,30000,24000,40000,4000],
            'Percentage':['10%','20%','25%','20%','10%','50%']
    })
```

```python
print('DF1: ', left_df)
print('DF2: ', right_df)
```

```python
#Pandas Function Merge
merged_df = pd.merge(left_df, right_df)
print(merged_df)
```

```python
merged_df = pd.merge(left_df, right_df, on=['Courses', 'Fee'])
print(merged_df)
```

```python
# Connecting df with differing column names
left_df = pd.DataFrame(
    {
        'Courses': ["Spark","PySpark","Python","pandas","Java"],
        'Price' : [20000,25000,30000,24000,40000],
        'Duration':['30day','40days','60days','55days','50days']
    })

right_df = pd.DataFrame(
    {
            'Class': ["Java","PySpark","Python","pandas","Hyperion","html"],
            'Fee': [20000,25000,30000,24000,40000,4000],
            'Percentage':['10%','20%','25%','20%','10%','50%']
    })
```

```python
result = pd.merge(left_df, right_df, left_on=['Courses', 'Price'], right_on=['Class', 'Fee'])

result
```

### .join() JOIN on indices or columns similar to 

```python
technologies = {
    'Courses':["Spark","PySpark","Python","pandas"],
    'Fee' :[20000,25000,22000,30000],
    'Duration':['30days','40days','35days','50days'],
    }
index_labels=['r1','r2','r3','r4']

df1 = pd.DataFrame(technologies,index=index_labels)
print("First DataFrame:\n", df1)

technologies2 = {
    'Courses':["Spark","Java","Python","Go"],
    'Discount':[2000,2300,1200,2000]
    }
index_labels2=['r1','r6','r3','r5']

df2 = pd.DataFrame(technologies2,index=index_labels2)
print("Second DataFRame:\n", df2)
```

```python
df1.join(df2, lsuffix='_left', rsuffix='_right')
```

```python
#df.join(other, lsuffix="_caller", rsuffix="_other")
#by default it joins using 
df1.join(df2, lsuffix='_left', rsuffix='_right', how='right')
```

```python
df1.join(df2, lsuffix='_left', rsuffix='_right', how='inner')
```

```python
# use merge in this case to join on a column this is another way to join
df1.set_index('Courses').join(df2.set_index('Courses'), lsuffix='_left', rsuffix='_right')
```

```python
# Expanded Sample Dataset (20 rows)
data = {
    'Department': [
        'Sales',       'Sales',        'Sales',        'Sales',        'Sales',        'Sales',        'Engineering',        'Engineering',        'Engineering',        'Engineering',
        'Engineering',        'HR',        'HR',        'HR',        'HR',        'Marketing',        'Marketing',        'Marketing',        'Marketing',        'Marketing',
    ],
    'Employee': [
        'Alice',        'Bob',        'Charlie',        'David',        'Eva',        'Frank',        'Grace',        'Hannah',        'Ian',        'Jack',
        'Karen',        'Liam',        'Mona',        'Nathan',        'Olivia',        'Paul',       'Quinn',        'Rachel',        'Sam',        'Tina',
        ],
    'Salary': [
        70000,        80000,        75000,        82000,        68000,        79000,        110000,        120000,        115000,        105000,        125000,        65000,
        60000,        62000,        67000,        72000,        78000,        85000,        69000,        74000,
        ],
        'Bonus': [        5000,        8000,        6000,        8500,        4500,        7500,        12000,        15000,        13000,        11000,        16000,
        4000,        3000,        3500,        4200,        5500,        7000,        9000,        5000,        6000,
    ],
    'Performance_Score': [        88,        92,        85,        90,        82,        89,        95,        98,        94,        91,        99,        78,        82,
        80,        84,        87,        91,        93,        83,        88,    ],
}

df = pd.DataFrame(data)
df.head(20)
```

```python
#TASK 1: calc total, avg, max values for salary in one command (single line)
df[['Salary']].agg(['sum', 'mean', 'max'])
```

```python
#TASK 2: calc min, max and avg across salary and bonus cols (single line)
df[['Salary', 'Bonus']].agg(['min', 'max', 'mean'])
```

```python
# TASK 3: calc specific metrics for ea col
#3.1 Salary - mean, median per dept
df.groupby(['Department'])[['Salary']].agg(['mean', 'median'])
```

```python
#3.2 Performance - highest per dept
#df.groupby(['Department'])[['Performance_Score']].max()

#option 2 places the data into a dictionary and pulls from that be sure to include curly braces { }
df.groupby('Department').agg({
    'Salary': ['mean', 'median'],
    'Performance_Score' : ['max'],
    'Bonus': ['sum']
})
```

```python
#3.3 Bonus - total paid per dept
df.groupby(['Department'])[['Bonus']].sum()
```

```python
#TASK 4: Calc avg salary, highest performance score, total bonus paid per dept
df.groupby('Department').agg ({
    'Salary': ['mean'],
    'Performance_Score': ['max'],
    'Bonus': ['sum']
})
```
